```python
import pandas as pd

data = pd.read_csv(r"F:\AnaPython\Superstore_Sales_Dataset.csv")

print(data.head())
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
# categorical columns
data['Country'] = data['Country'].fillna(data['Country'].mode()[0])
data['Region'] = data['Region'].fillna(data['Region'].mode()[0])
data['Category'] = data['Category'].fillna(data['Category'].mode()[0])

# numeric columns
data['Sales'] = data['Sales'].fillna(data['Sales'].mean())
data['Profit'] = data['Profit'].fillna(data['Profit'].mean())
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
```

```python
data.duplicated().sum()
```

```python
data['Country'] = data['Country'].str.upper()
```

```python
data.dtypes
```

```python
data.columns = data.columns.str.replace(" ", "")
```

```python
data.dtypes
```

```python
data['OrderDate'] = pd.to_datetime(data['OrderDate'], errors='coerce', dayfirst=True)
```

```python
data.info()
```

```python
print(data.head())
```

```python
data['Customer'] = data['Customer'].str.strip()
data['Region'] = data['Region'].str.strip()
data['Country'] = data['Country'].str.strip()
data['Category'] = data['Category'].str.strip()
```

```python
data['Profit_Margin'] = data['Profit'] / data['Sales']
```

```python
data[['Sales','Profit','Discount']] = data[['Sales','Profit','Discount']].round(2)
```

```python
data.info()
```

```python
print(data.head())
```

```python
data.to_csv("Cleaned_Sales_Dataset.csv", index=False)
```
